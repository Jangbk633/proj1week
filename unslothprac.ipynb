```python
from openai import OpenAI
model="kenonix/kanana-nano-2.1b-instruct-Q8_0-GGUF"
# 1. Unsloth 로컬 API 클라이언트 설정
client = OpenAI(
    base_url="http://127.0.0.1:8889/v1",
    api_key="sk-unsloth-f65f99d826c89c73362ddeb617e7963e",
)

# 2. 현재 Unsloth 서버에서 인식 중인 모델 ID 목록 확인
models = client.models.list()
print("사용 가능한 모델 목록:", [m.id for m in models.data])

# 3. 채팅 완료(Chat Completions) 요청
response = client.chat.completions.create(
    model=model,
    messages=[
        {"role": "system", "content": "너는 유용한 AI 어시스턴트야."},
        {"role": "user", "content": "Unsloth 모델을 API로 쓰는 법을 알려줘."},
    ],
    temperature=0.0,
)

print("\n[모델 응답]")
print(response.choices[0].message.content)
```

```python
stream = client.chat.completions.create(
    model=model,
    messages=[
        {"role": "system", "content": "너는 유용한 AI 어시스턴트야."},
        {"role": "user", "content": "파이썬의 장점을 3가지만 요약해줘."},
    ],
    temperature=0.7,
    stream=True,  # 스트리밍 활성화
)

print("[스트리밍 응답]")
for chunk in stream:
    if chunk.choices[0].delta.content is not None:
        print(chunk.choices[0].delta.content, end="", flush=True)
```
