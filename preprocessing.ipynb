```python
import sqlite3
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA, TruncatedSVD
from sklearn.manifold import TSNE

# 1. 윈도우 환경 한글 및 한자 호환 폰트 설정 (맑은 고딕)
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 2. SQLite3 데이터베이스에서 NEWS 테이블 로드
db_path = 'news.db'
conn = sqlite3.connect(db_path)
df = pd.read_sql_query("SELECT PK, FK, TITLE, REGDATE FROM NEWS", conn)
conn.close()

# 결측치 및 공백 데이터 제거
df = df.dropna(subset=['TITLE']).reset_index(drop=True)
df = df[df['TITLE'].str.strip() != ''].reset_index(drop=True)
print(f"총 로드된 뉴스 개수: {len(df)}개")

# 2-1. 한자 -> 한글 변환 딕셔너리 (뉴스 빈출 한자 매핑)
hanja_dict = {
    '北': '북', '中': '중', '美': '미', '日': '일', '露': '러', '韓': '한', '英': '영', '佛': '불', '獨': '독', '臺': '대',
    '尹': '윤', '李': '이', '文': '문', '朴': '박', '金': '김', '安': '안', '洪': '홍', '吳': '오', '曺': '조',
    '與': '여', '野': '야', '政': '정', '黨': '당', '靑': '청', '檢': '검', '警': '경', '軍': '군', '法': '법', '院': '원',
    '企': '기', '銀': '은', '證': '증', '株': '주', '債': '채', '稅': '세',
    '大': '대', '小': '소', '新': '신', '舊': '구', '高': '고', '低': '저', '多': '다', '少': '소',
    '前': '전', '後': '후', '上': '상', '下': '하', '內': '내', '外': '외', '左': '좌', '右': '우',
    '正': '정', '副': '부', '親': '친', '反': '반', '非': '비', '汎': '범', '全': '전',
    '萬': '만', '億': '억', '兆': '조', '年': '년', '月': '월', '時': '시', '分': '분', '秒': '초',
    '對': '대', '發': '발', '行': '행', '故': '고', '無': '무', '有': '유', '重': '중', '輕': '경'
}
hanja_trans = str.maketrans(hanja_dict)

# 2-2. 뉴스 상투어 및 기본 한국어 불용어 목록
stopwords = {
    # 뉴스 상투어
    '종합', '단독', '속보', '포토', '사진', '영상', '일문일답', '전문', '사설', '칼럼', '단신', 
    '브리핑', '초점', '현장', '르포', '인터뷰', '기획', '연재', '긴급', '주목', '화제', '이슈', '해설', '뉴스',
    # 빈출 시간/수량/의존 명사 및 대명사 불용어
    '만에', '하루', '연중', '내내', '이후', '부터', '까지', '위해', '대해', '통해', '따라', '관련', 
    '대한', '이번', '지난', '오늘', '내일', '어제', '올해', '내년', '작년', '현재', '지금', '당시',
    '우리', '저희', '그것', '이것', '저것', '여기', '거기', '저기', '모두', '전체', '일부', '각각',
    # 빈출 보조 용언 및 기타 불용 어절
    '있는', '없는', '하는', '되는', '않는', '같은', '위한', '통한', '따른', '대비', '두고', '놓고', 
    '있어', '없어', '하며', '되며', '하고', '되고', '한다', '된다', '했다', '됐다', '할까', '될까'
}

def preprocess_title(text):
    if not isinstance(text, str):
        return ""
    
    # (1) 한자 -> 한글 변환
    text = text.translate(hanja_trans)
    
    # (2) 정규표현식으로 대괄호[...], 소괄호(...), 꺾쇠<...>, 【...】 안의 상투어 패턴 제거
    tag_pattern = r'[\[\(\<【]\s*(?:종합|단독|속보|포토|사진|영상|일문일답|전문|사설|칼럼|브리핑|현장|르포|인터뷰|기획|긴급|뉴스|그래픽|카드뉴스|팩트체크|이슈)[0-9보\s]*[\]\)\>】]'
    text = re.sub(tag_pattern, ' ', text)
    
    # (3) 특수기호 정리 (한글, 영문, 숫자, 공백만 유지)
    text = re.sub(r'[^가-힣a-zA-Z0-9\s]', ' ', text)
    
    # (4) 어절 단위 분리 후 불용어 제거
    words = text.split()
    cleaned_words = [w for w in words if w not in stopwords]
    
    return " ".join(cleaned_words)

df['CLEAN_TITLE'] = df['TITLE'].apply(preprocess_title)
df = df[df['CLEAN_TITLE'].str.strip() != ''].reset_index(drop=True)
print(f"전처리 완료 후 유효 뉴스 개수: {len(df)}개")

# 2-3. 기존 데이터베이스(news.db)의 NEWS 테이블에 CLEAN_TITLE 컬럼 추가 및 저장
conn = sqlite3.connect(db_path)
cursor = conn.cursor()

cursor.execute("PRAGMA table_info(NEWS)")
existing_columns = [col[1] for col in cursor.fetchall()]

if 'CLEAN_TITLE' not in existing_columns:
    cursor.execute("ALTER TABLE NEWS ADD COLUMN CLEAN_TITLE TEXT")
    print("news.db의 NEWS 테이블에 'CLEAN_TITLE' 컬럼을 새로 추가했습니다.")

update_data = list(zip(df['CLEAN_TITLE'], df['PK']))
cursor.executemany("UPDATE NEWS SET CLEAN_TITLE = ? WHERE PK = ?", update_data)
conn.commit()
conn.close()
print("기존 news.db에 CLEAN_TITLE 데이터 업데이트를 완료했습니다.")
display(df[['PK', 'TITLE', 'CLEAN_TITLE']].head(5))

# 3. 텍스트 벡터화 (TF-IDF) - 전처리된 CLEAN_TITLE 사용
word_vectorizer = TfidfVectorizer(
    token_pattern=r'(?u)\b[가-힣a-zA-Z0-9]{2,}\b',
    max_features=3000
)
X_word = word_vectorizer.fit_transform(df['CLEAN_TITLE'])
feature_names = word_vectorizer.get_feature_names_out()

# 4. 최적의 n_clusters 탐색 (k = 2 ~ 100)
k_range = list(range(2, 101))
inertias = []
silhouette_scores = []

print("k=2~100 범위에서 최적의 클러스터 개수 탐색 중...")
for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_word)
    inertias.append(km.inertia_)
    silhouette_scores.append(silhouette_score(X_word, labels))

# [기준 A] 실루엣 점수가 가장 높은 k
k_A = k_range[int(np.argmax(silhouette_scores))]

# [기준 B] 엘보우 기법: 양 끝점(k=2, k=100)을 이은 직선으로부터 수직 거리가 가장 먼 k (정규화 기준)
x_norm = (np.array(k_range) - k_range[0]) / (k_range[-1] - k_range[0])
y_norm = (np.array(inertias) - min(inertias)) / (max(inertias) - min(inertias))

x1, y1 = x_norm[0], y_norm[0]
x2, y2 = x_norm[-1], y_norm[-1]

distances = np.abs((x2 - x1) * (y1 - y_norm) - (x1 - x_norm) * (y2 - y1)) / np.hypot(x2 - x1, y2 - y1)
k_B = k_range[int(np.argmax(distances))]

# [기준 C] A와 B의 평균 (반올림)
k_C = int(round((k_A + k_B) / 2))

print("\n[최적 클러스터 개수 도출 결과]")
print(f" - 기준 A (실루엣 점수 최대): k = {k_A} (점수: {max(silhouette_scores):.4f})")
print(f" - 기준 B (엘보우 곡률 최대): k = {k_B} (Inertia: {inertias[k_range.index(k_B)]:.2f})")
print(f" - 기준 C (A와 B의 평균)   : k = {k_C}")

# 4-1. 엘보우 및 실루엣 점수 그래프 시각화
fig, ax_metric = plt.subplots(1, 2, figsize=(16, 5))

ax_metric[0].plot(k_range, inertias, 'bo-', markersize=3, label='Inertia')
ax_metric[0].axvline(x=k_A, color='red', linestyle='--', label=f'A (실루엣 최적): k={k_A}')
ax_metric[0].axvline(x=k_B, color='green', linestyle='--', label=f'B (엘보우 최적): k={k_B}')
ax_metric[0].axvline(x=k_C, color='purple', linestyle=':', linewidth=2, label=f'C (평균): k={k_C}')
ax_metric[0].set_title('Elbow Method (Inertia 변화, k=2~100)', fontsize=13, fontweight='bold')
ax_metric[0].set_xlabel('Number of Clusters (k)')
ax_metric[0].set_ylabel('Inertia')
ax_metric[0].legend()
ax_metric[0].grid(True, linestyle='--', alpha=0.4)

ax_metric[1].plot(k_range, silhouette_scores, 'ro-', markersize=3, label='Silhouette Score')
ax_metric[1].axvline(x=k_A, color='red', linestyle='--', label=f'A (실루엣 최적): k={k_A}')
ax_metric[1].axvline(x=k_B, color='green', linestyle='--', label=f'B (엘보우 최적): k={k_B}')
ax_metric[1].axvline(x=k_C, color='purple', linestyle=':', linewidth=2, label=f'C (평균): k={k_C}')
ax_metric[1].set_title('Silhouette Analysis (실루엣 점수 변화, k=2~100)', fontsize=13, fontweight='bold')
ax_metric[1].set_xlabel('Number of Clusters (k)')
ax_metric[1].set_ylabel('Silhouette Score')
ax_metric[1].legend()
ax_metric[1].grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()

# 5. 3가지 차원 축소(PCA, TruncatedSVD, t-SNE) 사전 계산 (1회 수행 후 재사용)
print("\n차원 축소(PCA, TruncatedSVD, t-SNE) 계산 중...")
X_dense = X_word.toarray()
pca = PCA(n_components=2, random_state=42)
coords_pca = pca.fit_transform(X_dense)

svd = TruncatedSVD(n_components=2, random_state=42)
coords_svd = svd.fit_transform(X_word)

perplexity_val = min(30, max(5, len(df) - 1))
tsne = TSNE(n_components=2, random_state=42, perplexity=perplexity_val, init='random', learning_rate='auto')
coords_tsne = tsne.fit_transform(X_word)

methods = [
    ('PCA', coords_pca),
    ('TruncatedSVD', coords_svd),
    ('t-SNE', coords_tsne)
]

# 6. A, B, C 각 기준별 클러스터링 수행 및 산점도 시각화
criteria_list = [
    ('A (실루엣 최적)', k_A),
    ('B (엘보우 최적)', k_B),
    ('C (A·B 평균)', k_C)
]

top_n_keywords = 3
top_n_summary = 30
top_n_legend = 10
cmap = plt.get_cmap('tab20')

for crit_name, n_clusters in criteria_list:
    print("=" * 80)
    print(f"[{crit_name}] n_clusters = {n_clusters} 클러스터링 수행 결과")
    print("=" * 80)
    
    kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
    df['CLUSTER_ID'] = kmeans.fit_predict(X_word)
    
    centroids = kmeans.cluster_centers_
    cluster_labels = {}
    
    for i in range(n_clusters):
        top_indices = centroids[i].argsort()[::-1][:top_n_keywords]
        keywords = [feature_names[idx] for idx in top_indices if centroids[i][idx] > 0]
        if keywords:
            cluster_labels[i] = f"C{i}: " + ", ".join(keywords)
        else:
            cluster_labels[i] = f"C{i}: 기타"
            
    df['CLUSTER_LABEL'] = df['CLUSTER_ID'].map(cluster_labels)
    
    print(f"\n[클러스터별 할당 개수 및 핵심 키워드 라벨 (상위 {min(top_n_summary, n_clusters)}개)]")
    print(df['CLUSTER_LABEL'].value_counts().head(top_n_summary))
    display(df[['PK', 'TITLE', 'CLEAN_TITLE', 'CLUSTER_LABEL']].head(5))
    
    sorted_cluster_ids = df['CLUSTER_ID'].value_counts().index.tolist()
    top_10_cluster_ids = sorted_cluster_ids[:top_n_legend]
    
    fig, axes = plt.subplots(1, 3, figsize=(20, 6))
    fig.suptitle(f'[{crit_name}] 클러스터 시각화 (n_clusters = {n_clusters})', fontsize=16, fontweight='bold', y=1.02)
    
    for ax, (name, coords) in zip(axes, methods):
        for i in sorted_cluster_ids[top_n_legend:]:
            idx = (df['CLUSTER_ID'] == i)
            ax.scatter(
                coords[idx, 0], 
                coords[idx, 1], 
                color=cmap(i % 20),
                label='_nolegend_',
                alpha=0.5, 
                s=35,
                edgecolors='w',
                linewidth=0.5
            )
        for i in top_10_cluster_ids:
            idx = (df['CLUSTER_ID'] == i)
            count_val = idx.sum()
            ax.scatter(
                coords[idx, 0], 
                coords[idx, 1], 
                color=cmap(i % 20),
                label=f"{cluster_labels[i]} ({count_val}건)", 
                alpha=0.8, 
                s=45,
                edgecolors='w',
                linewidth=0.5
            )
            
        ax.set_title(f'{name} 기반 시각화', fontsize=14, fontweight='bold')
        ax.set_xlabel('Component 1')
        ax.set_ylabel('Component 2')
        ax.grid(True, linestyle='--', alpha=0.4)
        
    handles, labels = axes[0].get_legend_handles_labels()
    fig.legend(
        handles, labels, 
        loc='center left', 
        bbox_to_anchor=(1.0, 0.5), 
        title=f"상위 {min(top_n_legend, n_clusters)}개 클러스터 라벨", 
        fontsize=10, 
        title_fontsize=11
    )
    
    plt.tight_layout()
    plt.show()
```

```python
import sqlite3
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA, TruncatedSVD
from sklearn.manifold import TSNE

# 1. 윈도우 환경 한글 및 한자 호환 폰트 설정 (맑은 고딕)
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 2. SQLite3 데이터베이스에서 NEWS 테이블 로드
db_path = 'news.db'
conn = sqlite3.connect(db_path)
df = pd.read_sql_query("SELECT PK, FK, TITLE, REGDATE FROM NEWS", conn)
conn.close()

# 결측치 및 공백 데이터 제거
df = df.dropna(subset=['TITLE']).reset_index(drop=True)
df = df[df['TITLE'].str.strip() != ''].reset_index(drop=True)
print(f"총 로드된 뉴스 개수: {len(df)}개")

# 2-1. 의미 기반 한자 -> 한글 변환 규칙 설정
contextual_hanja_patterns = [
    # 직함이 결합된 인물 표기
    (r'李\s*대통령', '이재명 대통령 '),
    (r'李\s*대표', '이재명 대표 '),
    (r'尹\s*대통령', '윤석열 대통령 '),
    (r'尹\s*전\s*대통령', '윤석열 전 대통령 '),
    (r'文\s*전\s*대통령', '문재인 전 대통령 '),
    (r'文\s*대통령', '문재인 대통령 '),
    (r'韓\s*대표', '한동훈 대표 '),
    (r'韓\s*위원장', '한동훈 위원장 '),
    (r'吳\s*시장', '오세훈 시장 '),
    (r'洪\s*시장', '홍준표 시장 '),
    (r'曺\s*대표', '조국 대표 '),
    (r'安\s*의원', '안철수 의원 '),
    
    # 정치/기관 복합 표기
    (r'與\s*野|與·野', '여당 야당 '),
    (r'黨\s*政\s*大|당\s*정\s*대', '당정대 '),
    (r'黨\s*政|黨·政', '당정 '),
    (r'行\s*安\s*부|행\s*안\s*부', '행정안전부 '),
    
    # 국가 의미 복합 패턴
    (r'韓\s*국', '한국'),
    (r'韓·美·日|韓美日', '한국 미국 일본 '),
    (r'韓·中·日|韓中日', '한국 중국 일본 '),
    (r'韓·美|韓美|한·美|韓·미', '한국 미국 '),
    (r'韓·日|韓日|한·日|韓·일', '한국 일본 '),
    (r'韓·中|韓中|한·中|韓·중', '한국 중국 '),
    (r'韓·露|韓露|한·露|韓·러', '한국 러시아 '),
    (r'美·北|美北|北·美|北美', '미국 북한 '),
    (r'美·中|美中|中·美|中美', '미국 중국 '),
    (r'美·日|美日|日·美|日美', '미국 일본 '),
    (r'北·中|北中|中·北|中北', '북한 중국 '),
    (r'北·露|北露|露·北|露北', '북한 러시아 '),
    (r'방韓|訪韓', '방한'),
    (r'내韓|來韓', '내한'),
    (r'주韓|駐韓', '주한'),
    (r'반韓|反韓', '반한'),
    (r'친韓|親韓', '친한'),
    (r'혐韓|嫌韓', '혐한'),
    (r'주한\s*美\s*군', '주한 미군 '),
    (r'(?<=[방내주반친혐대訪來駐反親嫌對])美', '미'),
    (r'(?<=[방내주반친혐대訪來駐反親嫌對])中', '중'),
    (r'(?<=[방내주반친혐대訪來駐反親嫌對])日', '일'),
    (r'(?<=[방내주반친혐대訪來駐反親嫌對])北', '북'),
    (r'(?<=[방내주반친혐대訪來駐反親嫌對])露', '러'),
]

semantic_hanja_dict = {
    # 인물
    '李': '이재명 ', '尹': '윤석열 ', '韓': '한동훈 ', '文': '문재인 ',
    '洪': '홍준표 ', '吳': '오세훈 ', '安': '안철수 ', '曺': '조국 ',
    # 국가
    '美': '미국 ', '北': '북한 ', '中': '중국 ', '日': '일본 ', '露': '러시아 ',
    '英': '영국 ', '佛': '프랑스 ', '獨': '독일 ', '臺': '대만 ',
    # 정치/기관
    '與': '여당 ', '野': '야당 ', '政': '정부 ', '檢': '검찰 ', '警': '경찰 ',
    '軍': '군대 ', '靑': '청와대 ', '法': '법원 ', '銀': '은행 ', '證': '증권 ',
    '企': '기업 ', '黨': '정당 ',
}

phonetic_hanja_dict = {
    '萬': '만', '億': '억', '兆': '조',
    '年': '년', '月': '월', '時': '시', '分': '분', '秒': '초',
    '大': '대', '小': '소', '新': '신', '舊': '구', '高': '고', '低': '저', '多': '다', '少': '소',
    '前': '전', '後': '후', '上': '상', '下': '하', '內': '내', '外': '외', '左': '좌', '右': '우',
    '正': '정', '副': '부', '親': '친', '反': '반', '非': '비', '汎': '범', '全': '전',
    '對': '대', '發': '발', '行': '행', '故': '고', '無': '무', '有': '유', '重': '중', '輕': '경',
    '株': '주', '債': '채', '稅': '세', '金': '금', '院': '원', '朴': '박'
}
phonetic_trans = str.maketrans(phonetic_hanja_dict)

# 2-2. 뉴스 상투어 및 기본 한국어 불용어 목록
stopwords = {
    '종합', '단독', '속보', '포토', '사진', '영상', '일문일답', '전문', '사설', '칼럼', '단신', 
    '브리핑', '초점', '현장', '르포', '인터뷰', '기획', '연재', '긴급', '주목', '화제', '이슈', '해설', '뉴스',
    '만에', '하루', '연중', '내내', '이후', '부터', '까지', '위해', '대해', '통해', '따라', '관련', 
    '대한', '이번', '지난', '오늘', '내일', '어제', '올해', '내년', '작년', '현재', '지금', '당시',
    '우리', '저희', '그것', '이것', '저것', '여기', '거기', '저기', '모두', '전체', '일부', '각각',
    '있는', '없는', '하는', '되는', '않는', '같은', '위한', '통한', '따른', '대비', '두고', '놓고', 
    '있어', '없어', '하며', '되며', '하고', '되고', '한다', '된다', '했다', '됐다', '할까', '될까'
}

def preprocess_title(text):
    if not isinstance(text, str):
        return ""
    
    # [단계 1] 대괄호/소괄호 상투어 태그 제거
    tag_pattern = r'[\[\(\<【]\s*(?:종합|단독|속보|포토|사진|영상|일문일답|전문|사설|칼럼|브리핑|현장|르포|인터뷰|기획|긴급|뉴스|그래픽|카드뉴스|팩트체크|이슈)[0-9보\s]*[\]\)\>】]'
    text = re.sub(tag_pattern, ' ', text)
    
    # [단계 2] 숫자+日(날짜) 패턴 음차 처리
    text = re.sub(r'(\d+)\s*日', r'\1일', text)
    
    # [단계 3] 문맥 결합 복합 한자어 변환
    for pattern, replacement in contextual_hanja_patterns:
        text = re.sub(pattern, replacement, text)
        
    # [단계 4] 단독 약칭 한자 -> 전체 명칭 변환
    for hanja, full_word in semantic_hanja_dict.items():
        text = re.sub(hanja + r'(?=[은는이가을를의에로과와도만]\b)', full_word.strip(), text)
        text = text.replace(hanja, full_word)
        
    # [단계 5] 숫자, 접두/접미사 한자 음차 변환
    text = text.translate(phonetic_trans)
    
    # [단계 6] 특수기호 정리
    text = re.sub(r'[^가-힣a-zA-Z0-9\s]', ' ', text)
    
    # [단계 7] 어절 분리 후 조사 제거 및 불용어 필터링
    words = text.split()
    cleaned_words = []
    for w in words:
        if w in stopwords:
            continue
        if len(w) >= 3:
            w_stem = re.sub(r'(에서|으로|에게|부터|까지|이라|라고|은|는|이|가|을|를|의|에|로|과|와|도|만)$', '', w)
            if len(w_stem) >= 2:
                w = w_stem
        if w not in stopwords and len(w) >= 2:
            cleaned_words.append(w)
            
    return " ".join(cleaned_words)

df['CLEAN_TITLE'] = df['TITLE'].apply(preprocess_title)
df = df[df['CLEAN_TITLE'].str.strip() != ''].reset_index(drop=True)
print(f"전처리 완료 후 유효 뉴스 개수: {len(df)}개")

# 2-3. 기존 데이터베이스(news.db)의 NEWS 테이블에 CLEAN_TITLE 컬럼 추가 및 저장
conn = sqlite3.connect(db_path)
cursor = conn.cursor()

cursor.execute("PRAGMA table_info(NEWS)")
existing_columns = [col[1] for col in cursor.fetchall()]

if 'CLEAN_TITLE' not in existing_columns:
    cursor.execute("ALTER TABLE NEWS ADD COLUMN CLEAN_TITLE TEXT")
    print("news.db의 NEWS 테이블에 'CLEAN_TITLE' 컬럼을 새로 추가했습니다.")

update_data = list(zip(df['CLEAN_TITLE'], df['PK']))
cursor.executemany("UPDATE NEWS SET CLEAN_TITLE = ? WHERE PK = ?", update_data)
conn.commit()
conn.close()
print("기존 news.db에 의미 복원된 CLEAN_TITLE 데이터 업데이트를 완료했습니다.")
display(df[['PK', 'TITLE', 'CLEAN_TITLE']].head(10))

# 3. 텍스트 벡터화 (TF-IDF) - 전처리된 CLEAN_TITLE 사용
word_vectorizer = TfidfVectorizer(
    token_pattern=r'(?u)\b[가-힣a-zA-Z0-9]{2,}\b',
    max_features=3000
)
X_word = word_vectorizer.fit_transform(df['CLEAN_TITLE'])
feature_names = word_vectorizer.get_feature_names_out()

# 4. 최적의 n_clusters 탐색 (k = 2 ~ 100)
k_range = list(range(2, 101))
inertias = []
silhouette_scores = []

print("k=2~100 범위에서 최적의 클러스터 개수 탐색 중...")
for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_word)
    inertias.append(km.inertia_)
    silhouette_scores.append(silhouette_score(X_word, labels))

# [기준 A] 실루엣 점수가 가장 높은 k
k_A = k_range[int(np.argmax(silhouette_scores))]

# [기준 B] 엘보우 기법: 양 끝점(k=2, k=100)을 이은 직선으로부터 수직 거리가 가장 먼 k
x_norm = (np.array(k_range) - k_range[0]) / (k_range[-1] - k_range[0])
y_norm = (np.array(inertias) - min(inertias)) / (max(inertias) - min(inertias))

x1, y1 = x_norm[0], y_norm[0]
x2, y2 = x_norm[-1], y_norm[-1]

distances = np.abs((x2 - x1) * (y1 - y_norm) - (x1 - x_norm) * (y2 - y1)) / np.hypot(x2 - x1, y2 - y1)
k_B = k_range[int(np.argmax(distances))]

# [기준 C] A와 B의 평균 (반올림)
k_C = int(round((k_A + k_B) / 2))

print("\n[최적 클러스터 개수 도출 결과]")
print(f" - 기준 A (실루엣 점수 최대): k = {k_A} (점수: {max(silhouette_scores):.4f})")
print(f" - 기준 B (엘보우 곡률 최대): k = {k_B} (Inertia: {inertias[k_range.index(k_B)]:.2f})")
print(f" - 기준 C (A와 B의 평균)   : k = {k_C}")

# 4-1. 엘보우 및 실루엣 점수 그래프 시각화
fig, ax_metric = plt.subplots(1, 2, figsize=(16, 5))

ax_metric[0].plot(k_range, inertias, 'bo-', markersize=3, label='Inertia')
ax_metric[0].axvline(x=k_A, color='red', linestyle='--', label=f'A (실루엣 최적): k={k_A}')
ax_metric[0].axvline(x=k_B, color='green', linestyle='--', label=f'B (엘보우 최적): k={k_B}')
ax_metric[0].axvline(x=k_C, color='purple', linestyle=':', linewidth=2, label=f'C (평균): k={k_C}')
ax_metric[0].set_title('Elbow Method (Inertia 변화, k=2~100)', fontsize=13, fontweight='bold')
ax_metric[0].set_xlabel('Number of Clusters (k)')
ax_metric[0].set_ylabel('Inertia')
ax_metric[0].legend()
ax_metric[0].grid(True, linestyle='--', alpha=0.4)

ax_metric[1].plot(k_range, silhouette_scores, 'ro-', markersize=3, label='Silhouette Score')
ax_metric[1].axvline(x=k_A, color='red', linestyle='--', label=f'A (실루엣 최적): k={k_A}')
ax_metric[1].axvline(x=k_B, color='green', linestyle='--', label=f'B (엘보우 최적): k={k_B}')
ax_metric[1].axvline(x=k_C, color='purple', linestyle=':', linewidth=2, label=f'C (평균): k={k_C}')
ax_metric[1].set_title('Silhouette Analysis (실루엣 점수 변화, k=2~100)', fontsize=13, fontweight='bold')
ax_metric[1].set_xlabel('Number of Clusters (k)')
ax_metric[1].set_ylabel('Silhouette Score')
ax_metric[1].legend()
ax_metric[1].grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()

# 5. 3가지 차원 축소(PCA, TruncatedSVD, t-SNE) 사전 계산 (1회 수행 후 재사용)
print("\n차원 축소(PCA, TruncatedSVD, t-SNE) 계산 중...")
X_dense = X_word.toarray()
pca = PCA(n_components=2, random_state=42)
coords_pca = pca.fit_transform(X_dense)

svd = TruncatedSVD(n_components=2, random_state=42)
coords_svd = svd.fit_transform(X_word)

perplexity_val = min(30, max(5, len(df) - 1))
tsne = TSNE(n_components=2, random_state=42, perplexity=perplexity_val, init='random', learning_rate='auto')
coords_tsne = tsne.fit_transform(X_word)

methods = [
    ('PCA', coords_pca),
    ('TruncatedSVD', coords_svd),
    ('t-SNE', coords_tsne)
]

# 6. A, B, C 각 기준별 클러스터링 수행 및 산점도 시각화
criteria_list = [
    ('A (실루엣 최적)', k_A),
    ('B (엘보우 최적)', k_B),
    ('C (A·B 평균)', k_C)
]

top_n_keywords = 3
top_n_summary = 30
top_n_legend = 10
cmap = plt.get_cmap('tab20')

for crit_name, n_clusters in criteria_list:
    print("=" * 80)
    print(f"[{crit_name}] n_clusters = {n_clusters} 클러스터링 수행 결과")
    print("=" * 80)
    
    kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
    df['CLUSTER_ID'] = kmeans.fit_predict(X_word)
    
    centroids = kmeans.cluster_centers_
    cluster_labels = {}
    
    for i in range(n_clusters):
        top_indices = centroids[i].argsort()[::-1][:top_n_keywords]
        keywords = [feature_names[idx] for idx in top_indices if centroids[i][idx] > 0]
        if keywords:
            cluster_labels[i] = f"C{i}: " + ", ".join(keywords)
        else:
            cluster_labels[i] = f"C{i}: 기타"
            
    df['CLUSTER_LABEL'] = df['CLUSTER_ID'].map(cluster_labels)
    
    print(f"\n[클러스터별 할당 개수 및 핵심 키워드 라벨 (상위 {min(top_n_summary, n_clusters)}개)]")
    print(df['CLUSTER_LABEL'].value_counts().head(top_n_summary))
    display(df[['PK', 'TITLE', 'CLEAN_TITLE', 'CLUSTER_LABEL']].head(5))
    
    sorted_cluster_ids = df['CLUSTER_ID'].value_counts().index.tolist()
    top_10_cluster_ids = sorted_cluster_ids[:top_n_legend]
    
    fig, axes = plt.subplots(1, 3, figsize=(20, 6))
    fig.suptitle(f'[{crit_name}] 클러스터 시각화 (n_clusters = {n_clusters})', fontsize=16, fontweight='bold', y=1.02)
    
    for ax, (name, coords) in zip(axes, methods):
        for i in sorted_cluster_ids[top_n_legend:]:
            idx = (df['CLUSTER_ID'] == i)
            ax.scatter(
                coords[idx, 0], 
                coords[idx, 1], 
                color=cmap(i % 20),
                label='_nolegend_',
                alpha=0.5, 
                s=35,
                edgecolors='w',
                linewidth=0.5
            )
        for i in top_10_cluster_ids:
            idx = (df['CLUSTER_ID'] == i)
            count_val = idx.sum()
            ax.scatter(
                coords[idx, 0], 
                coords[idx, 1], 
                color=cmap(i % 20),
                label=f"{cluster_labels[i]} ({count_val}건)", 
                alpha=0.8, 
                s=45,
                edgecolors='w',
                linewidth=0.5
            )
            
        ax.set_title(f'{name} 기반 시각화', fontsize=14, fontweight='bold')
        ax.set_xlabel('Component 1')
        ax.set_ylabel('Component 2')
        ax.grid(True, linestyle='--', alpha=0.4)
        
    handles, labels = axes[0].get_legend_handles_labels()
    fig.legend(
        handles, labels, 
        loc='center left', 
        bbox_to_anchor=(1.0, 0.5), 
        title=f"상위 {min(top_n_legend, n_clusters)}개 클러스터 라벨", 
        fontsize=10, 
        title_fontsize=11
    )
    
    plt.tight_layout()
    plt.show()
```
