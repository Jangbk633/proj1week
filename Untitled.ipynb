```python
import sqlite3
import re
from datetime import datetime
from time import sleep
from requests import request
from requests.compat import urlparse, urljoin
from bs4 import BeautifulSoup


class NaverNewsCrawler:
    """
    네이버 뉴스의 오늘 날짜 기사를 크롤링하여 SQLite DB에 저장하는 클래스
    """
    def __init__(self, db_path='news.db', max_depth=3, delay=1.0):
        # 기본 설정 초기화
        self.db_path = db_path
        self.max_depth = max_depth
        self.delay = delay
        
        # 도메인 제한 설정 (기존 코드 유지)
        self.whitelist = ['news.naver.com', 'n.news.naver.com']
        self.blacklist = ['']
        
        # 네이버 서버 차단 방지를 위한 User-Agent 헤더 설정
        self.headers = {
            'User-Agent': (
                'Mozilla/5.0 (Windows NT 10.0; Win64; x64) '
                'AppleWebKit/537.36 (KHTML, like Gecko) '
                'Chrome/120.0.0.0 Safari/537.36'
            )
        }
        
        # 오늘 날짜 포맷 생성
        # 1) URL 파라미터용 (YYYYMMDD)
        # 2) 기사 본문 내 날짜 확인용 (YYYY.MM.DD.)
        now = datetime.now()
        self.today_param = now.strftime('%Y%m%d')
        self.today_dot = now.strftime('%Y.%m.%d.')
        
        # 네이버 뉴스 6개 주요 섹션 코드 (정치, 경제, 사회, 생활/문화, 세계, IT/과학)
        self.sections = ['100', '101', '102', '103', '104', '105']
        
        # URL Pool, 방문 기록(seen) 리스트 및 저장된 뉴스 카운트 초기화
        self.urls = []
        self.seen = []
        self.saved_count = 0

    def init_db(self):
        """데이터베이스 스키마 구축 (LINK, NEWS 테이블 초기화)"""
        with sqlite3.connect(self.db_path) as con:
            cur = con.cursor()
            cur.executescript('''
                DROP TABLE IF EXISTS LINK;
                CREATE TABLE LINK(
                    PK INTEGER PRIMARY KEY,
                    DEPTH INTEGER NOT NULL DEFAULT 1,
                    REFERER TEXT NOT NULL DEFAULT '',
                    URL TEXT NOT NULL,
                    REGDATE DATETIME DEFAULT CURRENT_TIMESTAMP
                );

                DROP TABLE IF EXISTS NEWS;
                CREATE TABLE NEWS(
                    PK INTEGER PRIMARY KEY,
                    FK INTEGER NOT NULL,
                    TITLE TEXT NOT NULL,
                    CONTENT TEXT NOT NULL,
                    REGDATE DATETIME DEFAULT CURRENT_TIMESTAMP
                );
            ''')
            con.commit()

    def seed_urls(self):
        """오늘 날짜의 6개 섹션별 1페이지 목록 URL을 초기 URL Pool에 등록"""
        # 메인 페이지 등록
        self.urls.append((1, '', 'https://news.naver.com'))
        
        # 6개 섹션의 오늘 날짜 1페이지 URL 등록 (Depth 1로 시작하여 페이지네이션 및 기사 탐색)
        for sid1 in self.sections:
            list_url = (
                f'https://news.naver.com/main/list.naver'
                f'?mode=LSD&mid=sec&sid1={sid1}&date={self.today_param}&page=1'
            )
            self.urls.append((1, 'https://news.naver.com', list_url))

    def is_valid_domain(self, url):
        """도메인 화이트리스트/블랙리스트 검사"""
        netloc = urlparse(url).netloc
        if netloc in self.blacklist and netloc not in self.whitelist:
            return False
        return netloc in self.whitelist

    def save_link(self, depth, referer, url):
        """방문한 URL을 LINK 테이블에 중복 없이 저장"""
        with sqlite3.connect(self.db_path) as con:
            cur = con.cursor()
            cur.execute('SELECT * FROM LINK WHERE URL = ?', [url])
            if len(cur.fetchall()) == 0:
                cur.execute('''
                    INSERT INTO LINK(DEPTH, REFERER, URL)
                    VALUES(?, ?, ?)
                ''', [depth, referer, url])
                con.commit()

    def is_today_article(self, dom):
        """기사 본문의 입력일이 오늘 날짜(YYYY.MM.DD.)인지 확인"""
        date_elem = dom.select_one('span._ARTICLE_DATE_TIME, span.t11')
        if date_elem:
            # 요소의 텍스트나 data-date-time 속성에 오늘 날짜가 포함되어 있는지 검사
            date_text = date_elem.text.strip()
            data_attr = date_elem.attrs.get('data-date-time', '')
            if self.today_dot in date_text or self.today_param_dash() in data_attr:
                return True
            return False
        # 날짜 요소가 없는 경우, 오늘 날짜 목록에서 넘어온 기사이므로 True 반환
        return True

    def today_param_dash(self):
        """data-date-time 속성 비교용 포맷 (YYYY-MM-DD)"""
        return f"{self.today_param[:4]}-{self.today_param[4:6]}-{self.today_param[6:]}"

    def save_news(self, url, dom):
        """본문 영역(#contents)에서 제목과 본문 요소가 모두 존재할 때만 DB에 저장"""
        if dom.select_one('#contents'):
            title = dom.select_one('h2#title_area')
            content = dom.select_one('article#dic_area')
            
            # 두 요소가 모두 존재하고 오늘 날짜 기사인 경우에만 DB 저장 (예외 처리 강화)
            if title is not None and content is not None:
                if self.is_today_article(dom):
                    with sqlite3.connect(self.db_path) as con:
                        cur = con.cursor()
                        cur.execute('''
                            INSERT INTO NEWS(FK, TITLE, CONTENT)
                            VALUES((SELECT PK FROM LINK WHERE URL=? LIMIT 0,1), ?, ?)
                        ''', [url, title.text.strip(), content.text.strip()])
                        con.commit()
                    
                    # 저장 건수 증가 및 한 줄 갱신 출력
                    self.saved_count += 1
                    print(f"\r현재 저장된 뉴스 기사 수: {self.saved_count}건", end='', flush=True)

    def extract_links(self, dom, current_url):
        """DOM에서 탐색할 하위 링크들을 추출"""
        links = []
        
        # 1) 기존 코드의 메뉴 및 섹션 기사 제목 선택자 유지
        if dom.select_one('ul.Nlnb_menu_list'):
            links.extend(dom.select('.Nlnb_menu_list > li > a[href]'))
        if dom.select_one('.section_article'):
            links.extend(dom.select('a.sa_text_title'))
            
        # 2) 날짜별 목록 페이지(list.naver)의 기사 링크 및 페이지네이션 링크 추출
        if dom.select_one('div.list_body'):
            links.extend(dom.select('div.list_body a[href]'))
            
        # 페이지네이션(다음 페이지 번호) 링크 추출 (오늘 날짜 파라미터가 포함된 경우만)
        if dom.select_one('div.paging'):
            for page_a in dom.select('div.paging a[href]'):
                href = page_a.attrs.get('href', '')
                if f'date={self.today_param}' in href or 'date=' not in href:
                    links.append(page_a)
                    
        return links

    def enqueue_links(self, links, current_depth, current_url):
        """추출한 링크를 정규화(Normalization)하여 URL Pool에 추가"""
        seen_urls = {row[2] for row in self.seen}
        queued_urls = {row[2] for row in self.urls}

        for link in links:
            href = link.attrs.get('href', '')
            if not href:
                continue
                
            nurl = urljoin(current_url, href)
            
            # javascript:, mailto:, # 등 비정상 링크 제외
            if re.match('http', nurl) is None or re.match('#', href):
                continue
                
            # 목록 페이지(list.naver)인 경우 오늘 날짜가 아니면 탐색에서 제외
            if 'list.naver' in nurl and f'date={self.today_param}' not in nurl:
                continue

            # 중복 방문 및 도메인 검사
            if (nurl not in seen_urls and 
                nurl not in queued_urls and 
                self.is_valid_domain(nurl)):
                
                # 페이지네이션 링크(list.naver)는 같은 Depth(1)를 유지하여 오늘자 모든 페이지를 순회하도록 설정
                if 'list.naver' in nurl and 'page=' in nurl:
                    next_depth = current_depth
                else:
                    next_depth = current_depth + 1
                    
                if next_depth <= self.max_depth:
                    self.urls.append((next_depth, current_url, nurl))
                    queued_urls.add(nurl)

    def run(self):
        """크롤러 메인 실행 루프"""
        self.init_db()
        self.seed_urls()
        self.saved_count = 0
        print(f"\r현재 저장된 뉴스 기사 수: {self.saved_count}건", end='', flush=True)

        while self.urls:
            depth, referer, url = self.urls.pop(0)
            self.seen.append((depth, referer, url))

            # Depth 제한 확인
            if depth > self.max_depth:
                continue

            # 도메인 제한 확인
            if not self.is_valid_domain(url):
                continue

            # 서버 차단 방지를 위한 요청 간 딜레이 (1초)
            sleep(self.delay)

            try:
                resp = request(url=url, method='GET', headers=self.headers)
            except Exception as e:
                print(f"\n요청 실패 ({url}): {e}")
                continue

            # 응답 상태 코드 확인 및 500번대 에러 시 튜플 형태로 재시도 큐에 추가
            if resp.status_code != 200:
                if resp.status_code >= 500:
                    self.urls.append((depth, referer, url))
                continue

            # HTML 응답 파싱 및 데이터 처리
            content_type = resp.headers.get('content-type', '')
            if re.search('text/html', content_type):
                dom = BeautifulSoup(resp.text, 'html.parser')

                # 1. 방문한 링크 DB 저장
                self.save_link(depth, referer, url)

                # 2. 기사 본문 추출 및 DB 저장
                self.save_news(url, dom)

                # 3. 하위 링크 추출 및 큐 등록
                links = self.extract_links(dom, url)
                self.enqueue_links(links, depth, url)

        print(f"\n크롤링 완료. 총 {self.saved_count}건의 뉴스가 저장되었습니다.")


# 스크립트 실행부
if __name__ == '__main__':
    crawler = NaverNewsCrawler(db_path='news.db', max_depth=3, delay=1.0)
    crawler.run()
```
